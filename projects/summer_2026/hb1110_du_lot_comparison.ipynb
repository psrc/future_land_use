#### HB1110 du per lot comparison
compares du per lot with and without hb1110 rules applied. Ignores du per acre entirely.

```python
import pandas as pd

from future_land_use.util.pipeline import Pipeline

p = Pipeline('configs')
```

```python
# without hb1110
flu = p.get_table('flu_imputed')
flu = flu.loc[(flu.Res_Use == 1) & (flu.FloorMaxDU_lot>0)]
xwalk = p.get_table('parcel_plan_type_xwalk')
no_hb1110 = xwalk.merge(flu,on=['juris_zn'])
no_hb1110 = no_hb1110.groupby(['Juris','juris_zn'])['FloorMaxDU_lot'].sum()
```

```python
# with hb1110
flu_hb1110 = p.get_table('flu_imputed_hb_1110')
flu_hb1110 = flu_hb1110.loc[(flu_hb1110.Res_Use == 1) & (flu_hb1110.FloorMaxDU_lot>0)]
xwalk_hb1110 = p.get_table('parcel_plan_type_xwalk_with_overlays_hb_1110')
hb1110 = xwalk_hb1110.merge(flu_hb1110,on=['juris_zn'])
hb1110 = hb1110.groupby(['Juris','juris_zn'])['FloorMaxDU_lot'].sum()
```

```python
df_out = pd.DataFrame({'du_lot': no_hb1110, 'du_lot_hb1110': hb1110})
df_out = df_out.loc[df_out.du_lot>0]
df_out.to_csv(f'{p.get_output_path()}/hb1110_du_lot_comparison.csv')
```
